```python
import datasets
import transformers
from accelerate import Accelerator
from IPython.display import Audio
from tqdm import tqdm
from transformers.pipelines.pt_utils import KeyDataset
```

```python
device = Accelerator().device
model_id = "openai/whisper-large-v3-turbo"
pipeline = transformers.pipeline(task="automatic-speech-recognition", model=model_id, language="en", device=device)
print(f"Using {pipeline.device}")
pipeline("https://huggingface.co/datasets/Narsil/asr_dummy/resolve/main/mlk.flac", return_timestamps=True)
```
